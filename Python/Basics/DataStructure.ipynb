## List comprehension 
### We are going to use looping statement inside the list so that we can store the values inside the list in a iterable format and in list comprehension conditional statements are allowed 

## Syntax
### [expression for item in iterable if condition]


```python
# I need to store values from to 50 inside the list
list2 = [x for x in range(1, 51)]
list2
```

# Tuple 

### A tuple is an ordered, indexed and immutable collection in Python that can store multiple objects.

## Tuples can store data in a ordered way just like list but tuples are not mutable 

## A Python tuple is:

## Ordered → maintains insertion order
## Indexed → elements can be accessed using indexes
## Immutable → cannot modify elements after creation
## Allows duplicates
## Can contain different data types
## Can contain other tuples/lists
## Generally useful for fixed collections of values 

```python
a = tuple()
```

```python
type(a)
```

```python
values = (1,2,3,4,5)
values
```

## Tuples are immutable 

```python
values[1]
```

```python
values[1] = 8
```

```python
tuples_data = (1,2,3,True,"Hello")
```

```python
tuples_data
```

```python
## Tuples are going to be ordered or are going to be indexed 
tuples_data[0]
```

```python
## Tuples also supports slicing just like lists. We can slice the tuples to get a subset of the tuple.
tuples_data[1:]
```

```python
## Tuples packing 
## Python automatically creates a tuple.
student = "Sanskar", 17, 7.75 
print(student)
```

```python
## We cannot add or remove elements from a tuple. Tuples are immutable.
## Because all these operations modify the tuple, they are not allowed.
student.append(18)
student.remove(17)
student.pop()
```

```python
t = (1,2,3,[4,5])
```

```python
## We can modify the list inside the tuple 
t[3].append(6)
```

```python
t
```

## Tuple immutability means the references/elements of the tuple cannot be reassigned, but if an element refers to a mutable object, that object itself may still be modified.

# Why Are Tuples Generally Faster Than Lists?

### Tuples are immutable and generally have lower overhead than lists, so they can be slightly faster and more memory-efficient for fixed collections.
