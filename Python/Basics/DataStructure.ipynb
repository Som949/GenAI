# Conditional Statements 

```python
amit = int(input("Enter your height amit: "))
shiv = int(input("Enter your height shiv: "))
if amit>shiv:
    print("Amit is taller")
elif amit==shiv:
    print("Both have same height")
else:
    print("Shiv is taller")

```

```python
## Whatever year is leap year or not 
year = int(input("Enter the year: "))
if year%4==0:
    if year%100==0:
        if year%400:
            print("Current is a leap year")
        else:
            print("Current is not a leap year")
    else:
        print("Year is leap year")
else:
    print("Year is not a leap year")
    
```

## Functions increases readiblity, reusability, for orgainizing the code as well 

```python
def even_odd(n):
    if n%2==0:
        output = "Number is even"
    else:
        output = "Number is odd"
    return output
```

```python
even_odd(5)
```

## Parameter is the input we need used to have when we define the function 

## Arguments is the input that we used to pass when we call the function 

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)
```

```python
num = int(input("Enter a number to find its factorial: "))
result = factorial(num)
print(f"The factorial of {num} is {result}")
```

```python
def palindrome(s):
    string_value = s.lower().replace(" ", "")
    if string_value == string_value[::-1]:
        print("Current string is a palindrome")
    else:
        print("Current string is not a palindrome")
```

```python
palindrome("mam")
```
