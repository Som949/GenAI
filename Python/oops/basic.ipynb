### Multilevel Inheritance 
#### Properties or methods of the grandparent class can be inherited by parent class and attributes or properties of 
#### parent class can be inherited by the child class.

#### GrandParent
#### |
#### | 
#### Parent
#### |

```python
class Grandparent:
    def __init__(self, nationality, confidence, anger_issue):
        self.nationality = nationality
        self.confidence = confidence
        self.anger_issue = anger_issue

    def introduction(self):
        print(f"My nationality is {self.nationality} and my confidence level is {self.confidence}")
```

```python
class Parent(Grandparent):
    def __init__(self, nationality, confidence, anger_issue, profession):
        super().__init__(nationality, confidence, anger_issue)
        self.profession = profession

    def introduction(self):
        super().introduction()
        print(f"My profession is {self.profession}") 
```

```python
class Child(Parent):
    def __init__(self, nationality, confidence, anger_issue, profession, hobby):
        super().__init__(nationality, confidence, anger_issue, profession)
        self.hobby = hobby

    def introduction(self):
        super().introduction()
        print(f"My hobby is {self.hobby}")
```

```python
Papaji = Parent("Indian", "High", "Yes", "Police Officer")
Papaji.introduction()
```

# Encapsulation 
## It is a way of wrapping data(variables and methods) inside a capsule or a class as a single unit
## It restricts direct access to some of the objects component   

### Key concepts of encapsulation 
#### 1. Public Members 
#### 2. Protected Members 
#### 3. Private members 
#### 4. Getter and setter 

### Public members - Public members can be accessed from anywhere: inside or outside the class.
### It is accessible by derives class and public attribute is dircetly declared 


```python
class Student:
    def __init__(self):
        self.name = "Sanskar"  # Public variable

    def show(self):             # Public method
        print(self.name)

s = Student()
print(s.name)
s.show()
```

### Private members - Private members use a double underscore (__). Python applies name mangling, making accidental access from outside the class harder. It cannot be accessed outside the class and derived class 

```python
class BankAccount:
    def __init__(self):
        self.__balance = 10000

    def show_balance(self):
        print(self.__balance)

b = BankAccount()
b.show_balance()

# print(b.__balance)  # AttributeError
```

### Protected members - Protected members are intended to be used inside the class and its subclasses. Python uses a single underscore (_) as a naming convention. It cannot be accessed outside the class but can be accessed from derived class 

```python
class Parent:
    def __init__(self):
        self._salary = 50000

class Child(Parent):
    def show(self):
        print(self._salary)

c = Child()
c.show()
print(c._salary)  # Accessible, but discouraged externally
# _salary is not strictly protected. Python allows external access, but the underscore signals that it is intended for internal use.
```

### Getter and Setter - A getter reads a value, while a setter updates a value with validation. They help control access to an attribute.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    # Getter
    def get_balance(self):
        return self.__balance

    # Setter
    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Balance cannot be negative")

b = BankAccount(10000)

print(b.get_balance())

b.set_balance(15000)
print(b.get_balance())

b.set_balance(-500)
```

# Abstraction 
## It is the concept of hiding the implementation details and only showing the neccessary feature of the object. 
## It helps in reducing programming complexity and efforts. 

## Python supports abstraction using the abc module, which provides:
### - ABC — a base class for defining an abstract class.
### - @abstractmethod — a decorator used to declare a method that subclasses must implement. 

### How to achieve abstraction in python 
#### 1. Abstract Class - Class that cannot be instatntiated or that means for an abstract class we cant create objects of the class
#### 2. Abstract Method - Just because it cant accessed by the object of the abstract class itself thats why we have to call abstract method in our child class and its implementation is not provided inside the abstract class but it must be provided in the child class. 
#### Abstract method is defined by the - @abstractmethod 

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def start(self):
        print("The car starts with an ignition key")

    def stop(self):
        print("The car stops with break pedals")

class ElectricScooter(Vehicle):

    def start(self):
        print("It start with a power button")

    def stop(self):
        print("It stops with an electric brake")
```

```python
car = Car()
car.start()
```

```python
scooter = ElectricScooter()
scooter.stop()
```

### What is abstraction? -  Hiding implementation details and exposing essential functionality.
### How do you achieve abstraction in Python? -  Using the abc module, ABC, and @abstractmethod.
### What is an abstract class? -  A class that cannot be instantiated while it contains unimplemented abstract methods.
### What is an abstract method? -  A method declared with @abstractmethod that normally requires subclasses to implement it.
### Can an abstract class have a constructor and normal methods? - Yes.
### Can an abstract class have both abstract and concrete methods? - Yes.
#### What is the main advantage of abstraction? - It establishes a common interface, reduces dependency on implementation details, and makes code easier to extend.

# Magic Methods:--
## Magic Methods are also as Dunder Methods because they have double underscore before and after their name and they are special methods in python that allows us to customize how objects behave 

## These methods help us to define:--
### 1. Object Initialization(__init__)
### 2. String Representation (__str__, __repr__)
### 3. Arthemetic Operation (__add__, __sub__)
### 4. Comparision Opeartion (__eq__, __lt__)
### 5. Length, Indexing, iteration(__len__, __getitem__)


### Why use?
1. Make objects behave like built in type 
2. Improves readiblity by allowing basic operations op1 + op2
3. Simplifies code by replacing verbose functions calls with natural syntax 

```python
## __init__ 
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks


s1 = Student("Rahul", 90)

print(s1.name)
print(s1.marks)
```

```python
## __str__
# The __str__() method defines what gets displayed when you print an object. 
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"{self.name} scored {self.marks} marks"


s1 = Student("Rahul", 90)

print(s1)
```

```python
## __add__
# The __add__() method allows you to define the behavior of the + operator 
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value


n1 = Number(10)
n2 = Number(20)

print(n1 + n2)
```

```python
## __len__
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)


team = Team(["Aman", "Riya", "Karan"])

print(len(team))
```

```python
## __gt__
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"{self.name}: {self.marks}"

    def __gt__(self, other):
        return self.marks > other.marks


s1 = Student("Rahul", 85)
s2 = Student("Aman", 92)

print(s1)
print(s2)
print(s2 > s1)
```

# Operator Overloading 
## Operator overloading means giving a special meaning to an operator when it is used with objects of a user-defined class.

### Why use?
1. Make custom objects behave like built in type 
2. Improves code reusability by using natural syntax 

#### Operator ---> + ---> Method name ---> __add__(self, other)
#### Operator ---> - ---> Method name ---> __sub__(self, other)
#### Operator ---> * ---> Method name ---> __mul__(self, other)
#### Operator ---> / ---> Method name ---> __truediv__(self, other)
#### Operator ---> == ---> Method name ---> __eq__(self, other)

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def __sub__(self, other):
        return Number(self.value - other.value)

    def __mul__(self, other):
        return Number(self.value * other.value)

    def __truediv__(self, other):
        return Number(self.value / other.value)

    def __eq__(self, other):
        return self.value == other.value

n1 = Number(15)
n2 = Number(5)

print("Addition", (n1 + n2).value)
print("Subraction", (n1 - n2).value)
print("Multiplication", (n1 * n2).value)
```
