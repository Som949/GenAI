## We are trying to match up with real world using programming

## Class is the blueprint for creating objects and it will remain same for each and every objects for us.

```python
class Avengers:

    def __init__(self, name, weapon, nationality="American"):
        self.name = name
        self.weapon = weapon
        self.nationality = nationality

    def introduction(self):
        print(f"Hello my name is {self.name}")

    def weapon_avenger(self):
        print(f"I am carrying the weapon {self.weapon}")
    
```

```python
Ironman = Avengers("Ironman", "Repulsor Rays")          #Object
print(Ironman.introduction())
```

```python
Ironman.weapon_avenger()
```

```python
Ironman.weapon
```

```python
Captain_America = Avengers("Captain America", "Shield")          #Object
```

```python
Captain_America.introduction()
```

```python
Captain_America.weapon
```

```python
Thor = Avengers("Thor", "God of Thunder", "Asgardian")          #Object
```

```python
Thor.nationality
```

```python
class Cartoon:

    def __init__(self, name, show, catchphrase):
        self.name = name
        self.show = show
        self.catchphrase = catchphrase

    def introduction(self):
        print(f"Hello my name is {self.name} and my show is {self.show}")

    def speak(self):
        print(f"{self.name} says {self.catchphrase}")
    
        
```

```python
Doremon = Cartoon("Doremon", "Doremon show", "Nobita!")
```

```python
Doremon.introduction
```

```python
Doremon.introduction()
```

```python
Doremon.speak()
```

# Types of Constructors:--
## 1. Default Constructor
## 2. Parmatrized Constructor

### Default - It doesn't have parameter inside it and it is having only the self keyword inside it.
### It assigns a default value to all objects. 

```python
class Dog:
    def __init__(self):
        self.name = "Dogesh"
        self.breed = "German Shepherd"
        self.color = "Black and Brown"

    def introduction(self):
        print(f"This dog is {self.name} and its color is {self.color}")
```

```python
Dog1 = Dog()
```

```python
Dog1.introduction()
```

```python
Dog1.name
```

### Parametrized - It consist of specific values inside it so that every object can have specific value 
### associated with it. 

```python
class Dogs:
    def __init__(self, breed, color):
        self.breed = breed
        self.color = color

    def intro(self):
        print(f"My name is {self.breed} and my color is {self.color}")

    
```
