### What is Pandas

Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool,
built on top of the Python programming language.

```python
import numpy as np
import pandas as pd
```

### Pandas Series is a column of data in a table like Excel or Google sheets with labels called indexes attached to each values.

```python
pd.Series(data=[40,60,80], index=['Bread', 'Milk', 'Cheese'], name='Price List')
```

### Series is a 1D array that can hold any type of data like integers, float, strings(objects), boolean, Mixed types.

### Creating a series in multiple ways
1. List
2. numpy array
3. Dictionary
4. Custom index
5. Scaler value

```python
## Series from list
country = ['India', 'USA', 'UK', 'Germany', 'France']
pd.Series(country)
```

```python
batsman_run = [48, 67, 102, 137, 92, 81]
pd.Series(batsman_run, dtype='int32')
```

```python
## Series using numpy array
values = np.array([10, 20, 30, 40, 50])
pd.Series(values)
```

```python
## Dictionary to Series
## Dictionary is another data structure that store data in key-value pairs. We can convert a dictionary to a Series using the pd.Series() function.
## We don't have to explictly mention the index
data = {'maths': 10, 'science': 20, 'english': 30}
pd.Series(data)
```

```python
## custom index to series 
values = [10, 20, 30, 40, 50]
index = ['a', 'b', 'c', 'd', 'e']
pd.Series(values, index=index)
```
