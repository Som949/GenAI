```python
import numpy as np
import pandas as pd
```

```python
# Task-1
marks = pd.Series([78, 85, 90, 66, 72])
print(marks)
print(marks.dtype)

print(marks[0])
print(marks[-2:])
```

```python
# Task-2
marks += 5
print(marks)
marks -= 2
print(marks)
marks *= 1.05
print(marks)
marks /= 2
print(marks)
```

```python
# Task-3
marks = pd.Series([78, 85, 90, 66, 72])
print("Maximum marks: ", marks.max())
print("Minimum marks: ", marks.min())
print("Sum of marks: ", marks.sum())
print("Mean of marks: ", marks.mean())

passed = marks.apply(lambda x: x>=70)
print("Number of students who passed: ", passed.sum())
```

```python
# Task-4
students = pd.DataFrame({
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
})

print("First 3 rows")
print(students[0:3])
print("Last 2 rows")
print(students[-2:])
print("DataFrame shape: ", students.shape)
print(students.columns)
```

```python
# Task-5
print("----------Task-5----------")
print(students.info())
print(students.describe())
print(students.head())
print(students.tail())
sorted = students.sort_values(by='Marks', ascending=False)
sorted.reset_index()
print(sorted)
```

```python
# Task-6
above75 = students[students['Marks']>75]
print("Students who scored more than 75 marks: ")
print(above75)
math = students[students['Subject'] == 'Math']
print("Students who took Math: ")
print(math)
above_avg = students[students['Marks']>students['Marks'].mean()]
print("Students who scored above average: ")
print(above_avg)
failed = students[students['Marks']<70]
print("Students who failed: ")
print(failed)
```

```python
# Task-7
print(students.groupby('Subject')['Marks'].mean())
print(students.groupby('Subject')['Name'].count())
print(students.groupby('Subject')['Marks'].max())
```

```python
# Task-8
students.plot(kind='bar', x='Name', y='Marks', title='Student Marks')
```

```python
# Task-9
sales = pd.DataFrame({
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Revenue': [1200, 1500, 900, 2000, 1800]
})

print("Total revenue: ", sales['Revenue'].sum())
print("Average revenue: \n", sales.groupby('Day')['Revenue'].mean())
print("Highest revenue: \n", sales.groupby('Day')['Revenue'].max())
print("Revenue greater than average: \n", sales[sales['Revenue']
                                                >sales['Revenue'].mean()])
sales.plot(kind='line', x='Day', y='Revenue', title='Daily Revenue')

```
